```python
#asarray
import numpy as np
a=[10,20,30,40]
b=np.asarray(a,dtype=float,order='C')
b
```

```python
s=([10,20,30],[40,50,60])
v=np.asarray(s,dtype=float,order='C')#Here 'C' means row major order and 'F' means column major order
print(v)
```

```python
#nditer means N-Dimensional Iterator. It is used to iterate over an array.
import numpy as np
s=([10,20,30],[40,50,60])
v=np.asarray(s,dtype=float,order='C')#here 'C' means row major order 
for i in np.nditer(v):
    print(i)
```

```python
v=np.asarray(s,dtype=float,order='F')#here 'F' means column major order 
for i in np.nditer(v):
    print(i)
```

```python
#frombuffer :It is used to create an array from a binary data source.
import numpy as np
a=b"welcome to python"
np.frombuffer(a,dtype='S1')#Here 'S1' means string of length 1
```

```python
np.frombuffer(a,dtype='S1',count=7)#here  count=7 means it will take first 7 elements from the binary data source.
```

```python
np.frombuffer(a,dtype='S1',offset=8)#here  offset=8 means it will take elements from the binary data source starting from index 8.
```

```python
np.frombuffer(a,dtype='S1',count=7,offset=2)
```

```python
#fromiter means 1-Dimensional Iterator. It is used to iterate over an array.
a=[10,20,30,40]
np.fromiter(a,dtype=float,count=2)
```

```python
np.fromiter(a,dtype='S1',count=2)#S1 is byte string.
```
